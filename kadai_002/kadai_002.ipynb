```python
import requests
import pprint
from getpass import getpass

# APIキーを指定
api_key = getpass('APIキーを入力してください: ')

# 検索キーワードを変数search_wordに格納
search_word = input('検索キーワードを入力してください: ')

# YouTube Data APIのURL（動画検索用）
url = 'https://www.googleapis.com/youtube/v3/search'

# データを取得
response = requests.get(
    url,
    params={
        'key': api_key,  # APIキーをパラメータに追加
        'part': 'snippet',  # 動画情報の詳細を指定
        'q': search_word,  # 検索キーワードをパラメータに追加
        'type': 'video',  # 動画のみを検索対象に指定
        'maxResults': 10  # 取得する動画の最大数を指定
    }
)
# レスポンスのJSONを取得
json_data = response.json()

# JSONの中身を確認
pprint.pprint(json_data)

# 取得した動画のタイトルとURLを表示
for item in json_data['items']:
    video_title = item['snippet']['title']
    video_url = f'https://www.youtube.com/watch?v={item["id"]["videoId"]}'
    print(f'{video_title}: {video_url}')
```

```python

import requests
import pprint
from getpass import getpass

# ① APIキーを安全に入力
api_key = getpass('APIキーを入力してください: ')

# ② 駅名を入力
search_word = input('検索する駅を入力してください: ')

# ③ 駅の緯度・経度を取得
url = 'https://maps.googleapis.com/maps/api/geocode/json'

response = requests.get(
    url,
    params={
        'address': search_word,
        'key': api_key,
        'language': 'ja'
    },
    timeout=10
)

data = response.json()

if data.get('status') != 'OK' or not data.get('results'):
    print('駅の位置情報を取得できませんでした。')
    print(data)
else:
    location = data['results'][0]['geometry']['location']

    latitude = location['lat']
    longitude = location['lng']

    print('緯度:', latitude)
    print('経度:', longitude)


# ④ 半径500m以内のレストランを検索
url = 'https://places.googleapis.com/v1/places:searchNearby'

headers = {
    'X-Goog-Api-Key': api_key,
    'X-Goog-FieldMask': 'places.displayName,places.rating,places.formattedAddress',
    'Content-Type': 'application/json'
}

payload = {
    'includedTypes': ['restaurant'],
    'maxResultCount': 20,
    'locationRestriction': {
        'circle': {
            'center': {
                'latitude': latitude,
                'longitude': longitude
            },
            'radius': 500.0
        }
    },
    'languageCode': 'ja'
}

response = requests.post(
    url,
    headers=headers,
    json=payload,
    timeout=10
)

restaurant_data = response.json()

# ⑤ レストラン情報を表示
if response.status_code == 200:
    restaurants = restaurant_data.get('places', [])

    print(f'\n検索結果: {len(restaurants)}件')

    for restaurant in restaurants:
        print('--------------------')
        print('レストラン名:', restaurant.get('displayName', {}).get('text', '不明'))
        print('評価:', restaurant.get('rating', '評価なし'))
        print('住所:', restaurant.get('formattedAddress', '住所なし'))
else:
    print('レストラン情報を取得できませんでした。')
    pprint.pprint(restaurant_data)
```
